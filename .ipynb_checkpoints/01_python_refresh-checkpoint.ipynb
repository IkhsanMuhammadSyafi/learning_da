```python
# 1. Variables & Data Types
nama_aset = "Laptop Dell Latitude PSID"  # String (teks)
jumlah_unit = 15                          # Integer (bilangan bulat)
harga_satuan_juta = 12.5                  # Float (desimal)
is_aktif = True                           # Boolean (True/False)

# Hitung total nilai aset
total_nilai = jumlah_unit * harga_satuan_juta

print(f"Aset: {nama_aset}")
print(f"Total Nilai: Rp {total_nilai} Juta")
```

```python
# 2. List of Dictionaries (Simulasi Data Aset PSID)
daftar_aset = [
    {"nama": "Laptop Dell", "kategori": "Elektronik", "jumlah": 15, "harga_juta": 12.5},
    {"nama": "Meja Kerja", "kategori": "Furnitur", "jumlah": 20, "harga_juta": 1.8},
    {"nama": "Server Rak OJS", "kategori": "Elektronik", "jumlah": 2, "harga_juta": 45.0},
    {"nama": "Kursi Putar", "kategori": "Furnitur", "jumlah": 25, "harga_juta": 0.9}
]

print(f"Total jenis aset: {len(daftar_aset)}")
print("Aset pertama:", daftar_aset[0]["nama"])
```

```python
total_valuasi_semua = 0
total_valuasi_elektronik = 0

for aset in daftar_aset:
    nilai_aset = aset["jumlah"] * aset["harga_juta"]
    total_valuasi_semua += nilai_aset
    
    # Kondisi: hanya jika Elektronik
    if aset["kategori"] == "Elektronik":
        total_valuasi_elektronik += nilai_aset
        print(f"-> {aset['nama']}: Rp {nilai_aset:.1f} Juta")

print("---------------------------------")
print(f"Total Seluruh Aset     : Rp {total_valuasi_semua:.1f} Juta")
print(f"Total Aset Elektronik  : Rp {total_valuasi_elektronik:.1f} Juta")
```

```python
# 3. Membuat Fungsi Hitung Valuasi
def hitung_valuasi(jumlah, harga_per_unit):
    """Fungsi sederhana untuk menghitung total nilai aset"""
    total = jumlah * harga_per_unit
    return total

# Cara pakainya gampang banget:
laptop_psid = hitung_valuasi(15, 12.5)
server_ojs = hitung_valuasi(2, 45.0)

print(f"Valuasi Laptop: Rp {laptop_psid} Juta")
print(f"Valuasi Server: Rp {server_ojs} Juta")
```
